### 1.1. `np.array`: создание массива, `shape` и `dtype`

`np.array` создаёт числовой массив `ndarray` из списка или вложенных списков. Он нужен, когда над множеством чисел нужно выполнять одинаковые вычисления: например, обработать время и количество ингредиентов всех рецептов сразу. NumPy импортируют как `import numpy as np`; `np` — короткое имя библиотеки.

**Как записывать**

```python
a = np.array([1, 2, 3], dtype=np.int32)
m = np.array([[1, 2], [3, 4]], dtype=np.float64)
print(m.shape, m.ndim, m.size, m.dtype)
```

**Параметры и свойства**

| Запись | Что означает |
|---|---|
| `object` — первый аргумент | Исходные данные: список, кортеж, другой массив |
| `dtype=np.int32` | Хранить целые числа в 32 битах; можно выбрать `np.int64`, `np.float64`, `bool` |
| `ndmin=2` | Создать массив минимум с двумя измерениями; `[1,2]` станет строкой формы `(1,2)` |
| `copy=True` | Создать отдельные данные; это обычное поведение `np.array` |
| `a.shape` | Форма: `(3,)` — три числа, `(2,3)` — две строки по три числа |
| `a.ndim` | Количество осей: длина кортежа `shape` |
| `a.size` | Общее число элементов; для `(2,3)` это 6 |
| `a.dtype` | Тип каждого элемента массива |
| `a.itemsize`, `a.nbytes` | Байты на элемент и на все значения: `size * itemsize` |

**Как это работает**

Параметр передают в круглых скобках при вызове функции: `np.array(..., dtype=...)`. Свойство читают через точку без скобок: `a.shape`, не `a.shape()`. NumPy выбирает общий тип для всего массива: если среди чисел есть дробное, результат обычно будет вещественным. Вложенные списки должны быть прямоугольными: одинаковое число элементов в каждой строке.

**Частая ошибка:** Не путай `shape` с количеством элементов. В `(3,)` запятая обозначает кортеж из одного числа. Принудительное преобразование дробных чисел в целый тип может потерять дробную часть.

**Пример**

````python
import numpy as np
from IPython.display import display
np.set_printoptions(precision=3, suppress=True)

toy = np.array([[101, 10, 3], [102, 0, 5], [103, 30, 8]], dtype=np.int32)
print(toy)
print('shape:', toy.shape, 'ndim:', toy.ndim, 'size:', toy.size, 'dtype:', toy.dtype)
print('Байты:', toy.nbytes, '=', toy.size, '*', toy.itemsize)
````

### 1.2. Индексы и срезы: `a[строки, столбцы]`

Индексация нужна, когда из массива требуется получить конкретное число, столбец признака или часть таблицы. Индексы — позиции от нуля, отрицательные индексы отсчитываются с конца.

**Как записывать**

```python
a[0, 1]       # одно значение: первая строка, второй столбец
a[:, 1]       # весь второй столбец
a[:2, 1:]     # первые две строки, без первого столбца
a[::2, ::2]   # каждая вторая строка и каждый второй столбец
```

**Параметры и свойства**

- Срез записывается `start:stop:step`: начало, граница, шаг. `stop` не входит.
- `:` без границ означает все позиции. `-1` — последняя позиция.
- `a[:, 1]` убирает ось столбцов; `a[:, 1:2]` сохраняет её длиной 1.
- `a[[2,0]]` выбирает строки по списку индексов в указанном порядке.

**Как это работает**

Сначала выбираются строки (до запятой), затем столбцы (после неё). У массива формы `(3,3)` столбец `a[:,1]` имеет форму `(3,)`, а тот же столбец как таблица `a[:,1:2]` — `(3,1)`. Обычный срез часто работает с той же памятью; чтобы редактировать независимо, добавь `.copy()`.

**Частая ошибка:** `a[1]` — вторая строка, а не второй столбец. Изменение среза может изменить исходный массив.

**Пример**

````python
print('Первые 2 строки:\n', toy[:2])
print('Минуты:', toy[:, 1])
print('Без ID:\n', toy[:, 1:])
````

### 1.3. Булевы маски и `copy`: выбрать, посчитать, изменить

Маска нужна, чтобы отбирать строки по условиям без Python-цикла. Например, найти рецепты с ≤5 ингредиентами или исправить нулевую длительность.

**Как записывать**

```python
mask = a[:, 2] <= 5
selected = a[mask]
count = mask.sum()
share = mask.mean()
b = a.copy()
b[b[:, 1] == 0, 1] = 1
```

**Параметры и свойства**

- Сравнения: `<`, `<=`, `>`, `>=`, `==`, `!=` — возвращают массив `bool`.
- `&` — оба условия; `|` — хотя бы одно; `~mask` — отрицание.
- `.sum()` для маски считает `True`; `.mean()` вычисляет долю от 0 до 1.
- `.copy()` создаёт независимый массив.
- `a[mask, column] = value` меняет только указанные позиции.

**Как это работает**

В маске `True` означает «оставить». Длина маски строк должна совпадать с числом строк массива. Для каждого условия нужны скобки: `(a[:,1] <= 20) & (a[:,2] <= 5)`. Процент — `100 * mask.mean()`.

**Частая ошибка:** `and` и `or` работают с отдельными логическими значениями, а не выполняют поэлементное сравнение массивов. `b = a` не создаёт копию.

**Пример**

````python
mask = toy[:, 2] <= 5
print('Маска:', mask)
print('Подходящие рецепты:\n', toy[mask])
print('Количество:', mask.sum(), 'Доля:', mask.mean())
````

<details><summary>Подсказка — открой после своей попытки</summary>

`copy()` защищает исходные данные; присваивание: `копия[маска, 1] = 1`.

### 1.4. `arange`, `reshape` и агрегаты по `axis`

Агрегаты сводят много значений к статистике: сумме, среднему, минимуму, максимуму или медиане. Параметр `axis` определяет, считать по каждому столбцу, по каждой строке или по всему массиву.

**Как записывать**

```python
a = np.arange(12).reshape(3, 4)
a.sum(axis=0)
a.mean(axis=1)
np.median(a, axis=0)
```

**Параметры и свойства**

- `np.arange(start, stop, step)` создаёт последовательность без `stop`; `np.arange(12)` — числа 0…11.
- `.reshape(3,4)` задаёт новую форму; число элементов должно остаться тем же.
- `axis=None` (обычно по умолчанию) — один агрегат по всем значениям.
- `axis=0` — сворачиваем строки: одно значение на столбец.
- `axis=1` — сворачиваем столбцы: одно значение на строку.
- `keepdims=True` сохраняет свёрнутую ось размером 1.
- `.sum()`, `.mean()`, `.min()`, `.max()`, `np.median(...)` — разные агрегаты.

**Как это работает**

Для формы `(3,4)` результат по `axis=0` имеет форму `(4,)`, по `axis=1` — `(3,)`. Среднее суммирует и делит на число значений; медиана берёт середину отсортированных значений. Чтобы убрать ID из статистики, сначала выбери `a[:,1:]`. Для нормализации строк удобно `a / a.sum(axis=1, keepdims=True)` — если суммы не нулевые.

**Частая ошибка:** `reshape` не создаёт новые элементы. Правило «ось сворачивается» относится к агрегатам; `cumsum` сохраняет форму.

**Пример**

````python
values = toy_fixed[:, 1:]
print('Средние по столбцам:', values.mean(axis=0))
print('Минимумы:', values.min(axis=0))
print('Медианы:', np.median(values, axis=0))
````

### 1.5. `unique`, `argsort` и случайная выборка

Эти инструменты отвечают на разные вопросы: какие значения встречаются, в каком порядке идут строки и какие строки выбрать случайно. В таблице важно не потерять связь ID с его признаками.

**Как записывать**

```python
values, positions = np.unique(a[:,0], return_index=True)
order = np.argsort(a[:,1], kind="stable")
top = a[order[-2:]]
rng = np.random.default_rng(42)
chosen = rng.choice(len(a), size=2, replace=False)
```

**Параметры и свойства**

- `np.unique(a)` возвращает разные значения, по умолчанию отсортированные.
- `return_index=True` дополнительно возвращает позиции первых вхождений.
- `return_counts=True` возвращает число вхождений каждого значения.
- `argsort(..., axis=-1, kind="stable")` возвращает позиции в порядке возрастания; stable сохраняет порядок равных значений.
- `default_rng(seed)` создаёт генератор; одинаковый seed повторяет выбор.
- `choice(a, size, replace, p)` выбирает из числа позиций или готового массива; `p` — вероятности.
- `replace=False` запрещает повторный выбор одной позиции.

**Как это работает**

Сначала сортируй один столбец через `argsort`, затем выбирай целые строки по полученным позициям. Для самых больших значений берут последние позиции. Случайно выбирай из уникальных ID, если по условию нужны разные рецепты: разные строки не всегда означают разные ID.

**Частая ошибка:** `np.sort(a[:,1])` даёт отсортированные минуты, но не переставляет остальные столбцы согласованно. Нельзя выбрать больше разных позиций, чем есть строк.

**Пример**

````python
print('Разные количества ингредиентов:', np.unique(toy_fixed[:, 2]))
order = np.argsort(toy_fixed[:, 1])
print('Порядок строк:', order)
print('Два самых долгих:\n', toy_fixed[order[-2:]])
````

### 1.6. `np.loadtxt`: числовой CSV

`loadtxt` читает прямоугольную числовую таблицу из текста. Он нужен в лабораторной, чтобы загрузить сразу все рецепты в массив, а не разбирать строки вручную.

**Как записывать**

```python
data = np.loadtxt(path, delimiter=",", skiprows=1, dtype=np.int32, ndmin=2)
```

**Параметры и свойства**

- Первый аргумент — путь или текстовый поток.
- `delimiter` — разделитель; для CSV — запятая.
- `skiprows` — сколько строк в начале пропустить; 1 исключает шапку.
- `dtype` — числовой тип.
- `usecols=(0,2)` выбирает нужные столбцы.
- `ndmin=2` сохраняет двумерность даже для одной строки.

**Как это работает**

Заголовок содержит названия, а не числа, поэтому его пропускают. Все оставшиеся ячейки должны преобразовываться в выбранный dtype. Для маленького примера `StringIO` играет роль файла, но текст находится прямо в ячейке.

**Частая ошибка:** Если забыть skiprows, NumPy попытается превратить слово id в число. Для сложных смешанных таблиц удобнее Pandas.

**Пример**

````python
from io import StringIO
text_csv = "id,minutes,ingredients\n101,10,3\n102,20,5\n"
csv_demo = np.loadtxt(StringIO(text_csv), delimiter=",", skiprows=1, dtype=np.int32)
print(csv_demo)
````

### 1.7. `np.column_stack`: добавить признак

Новый вычисленный признак нужно сохранить рядом с исходными данными. `column_stack` собирает одинаковые по длине столбцы в одну таблицу.

**Как записывать**

```python
mask = a[:,1] <= 20
result = np.column_stack((a, mask.astype(np.int32)))
```

**Параметры и свойства**

- Единственный аргумент — последовательность массивов, например кортеж `(a, new_column)`.
- 1D-массив трактуется как столбец.
- Число строк всех частей должно совпадать.
- `.astype(dtype)` преобразует тип, например True/False в 1/0.

**Как это работает**

Исходный массив не растёт на месте: функция возвращает новый. Если типы частей различаются, выбирается общий тип результата. Бинарный столбец удобен тем, что его среднее равно доле единиц.

**Частая ошибка:** Скобки важны: `column_stack((a,b))`, а не два независимых позиционных аргумента.

**Пример**

````python
short_toy = (toy_fixed[:,1] <= 20).astype(np.int32)
print(np.column_stack((toy_fixed,short_toy)))
````

### 3.1. Broadcasting: столбец × строка

Broadcasting позволяет выполнять операции с массивами совместимых форм без явного размножения данных. Так создают таблицу умножения, вычитают среднее из каждого столбца или считают расстояния между точками.

**Как записывать**

```python
v = np.arange(1, 4)
column = v[:, None]
row = v[None, :]
table = column * row
```

**Параметры и свойства**

- `None` или `np.newaxis` добавляет ось размера 1.
- `v[:,None]` — форма `(N,1)`, `v[None,:]` — `(1,N)`.
- Формы сравниваются справа: соответствующие размеры должны совпадать или один должен быть равен 1.
- `*` здесь выполняет поэлементное умножение, не матричное.

**Как это работает**

В результате `(N,1) * (1,N)` получится `(N,N)`: каждая строка содержит множитель из столбца, умноженный на все числа строки. Сам вектор `v` имеет форму `(N,)`: он не является ни двумерной строкой, ни столбцом.

**Частая ошибка:** `.T` у одномерного вектора не делает столбец. Добавь отдельную ось явно.

**Пример**

````python
numbers = np.arange(1, 4)
print(numbers[:, None].shape, numbers[None, :].shape)
print(numbers[:, None] * numbers[None, :])
````

### 3.2. `np.diag` и арифметическая прогрессия

`np.diag` из вектора создаёт диагональную матрицу; из двумерной матрицы извлекает диагональ. Он нужен, когда значения должны располагаться только на диагонали: например, коэффициенты масштабирования или члены прогрессии.

**Как записывать**

```python
members = first + step * np.arange(N)
matrix = np.diag(members, k=0)
diagonal = np.diag(matrix)
```

**Параметры и свойства**

- `v` — вектор для построения либо матрица для извлечения диагонали.
- `k=0` — главная диагональ, `k=1` — выше, `k=-1` — ниже.
- `first` — первый член; `step` — разность; `N` — число членов.
- `np.arange(N)` даёт номера 0…N−1.

**Как это работает**

Нулевой номер даёт первый член `first`; следующий — `first + step`. Остальные элементы построенной матрицы равны нулю. Чтобы переиспользовать расчёт, заверни выражение в `def` и верни результат через `return`.

**Частая ошибка:** Не начинай номера с 1, иначе первый член станет `first + step`.

**Пример**

````python
progression = 1.5 + 0.5 * np.arange(4)
print(progression)
print(np.diag(progression))
````

### 3.5. `numpy.linalg`: определитель и обратная матрица

Линейная алгебра нужна для систем уравнений, преобразований и моделей. Перед поиском обратной матрицы важно понять, существует ли она.

**Как записывать**

```python
det = np.linalg.det(A)
rank = np.linalg.matrix_rank(A)
A_inverse = np.linalg.inv(A)
np.allclose(A @ A_inverse, np.eye(len(A)))
```

**Параметры и свойства**

- `A` для `det` и `inv` должна быть квадратной.
- `matrix_rank(A)` оценивает число независимых строк/столбцов.
- `inv(A)` вычисляет обратную; при вырожденности возможен `LinAlgError`.
- `@` — матричное умножение, `*` — поэлементное.
- `np.eye(N)` — единичная матрица; `allclose` сравнивает с допуском.

**Как это работает**

Если ранг меньше размера квадратной матрицы, обычной обратной матрицы нет. Для треугольной матрицы определитель — произведение диагональных значений. При численных вычислениях произведение `A @ inv(A)` может отличаться от единичной матрицы на малую погрешность.

**Частая ошибка:** Для решения `A @ x = b` обычно используют `np.linalg.solve(A,b)`, а не вычисляют `inv(A)` только ради умножения на b.

**Пример**

````python
invertible_demo = np.array([[2., 1.], [1., 3.]])
print('Детерминант:', np.linalg.det(invertible_demo))
print('Обратная:\n', np.linalg.inv(invertible_demo))
````
